## 1. Question

*The research question and the decision it supports.*
Search performance changes continuously because of evolving search algorithms, user behavior, competition, and content freshness. Identifying pages that may decline before substantial traffic loss occurs allows SEO and content teams to allocate effort more efficiently.

The objective of this project is to investigate whether machine learning models trained on historical SEO signals can rank pages according to their likelihood of future decline. Rather than automating business decisions, the system is designed to support human reviewers by highlighting pages that deserve additional attention.
