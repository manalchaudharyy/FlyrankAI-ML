**Decision this improves:** which pages a content team should review first when they only
have capacity to fix a limited number per cycle (e.g. 20-50 pages/month, not all 30,000).

**Who acts on it:** a content strategist or SEO reviewer, who takes the ranked queue and
decides refresh, expand, protect, prune, or monitor for each top-ranked page.

**Unit of analysis:** one page (one row in the starter dataset = one content_id).

**Cost of a wrong call:** if the queue is wrong, the team spends limited reviewer hours on
pages that were never actually declining (wasted effort), while pages that genuinely needed
attention keep losing visibility unnoticed (missed opportunity, real traffic/revenue cost).
Because this is a ranked recommendation feeding a human reviewer -- not an automated action --
a wrong call costs time, not irreversible damage. But time is the scarce resource here, so
precision at the top of the list matters more than a high score in isolation.


*Why data/ML helps at all:** a hand-written rule (stale + visible pages, sorted by
impressions) is a reasonable starting point, but Week 1-2 evidence already shows a learned
model finds patterns a fixed rule misses (3x precision lift). A simple rule can't weigh
multiple weak signals (age, position, CTR, word count) against each other the way a model can.

## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
import os, sys, subprocess

IN_COLAB = "google.colab" in sys.modules
REPO_URL = "https://github.com/manalchaudharyy/FlyrankAI-ML"
REPO_DIR = "FlyrankAI-ML"

if IN_COLAB:
    if not os.path.isdir(REPO_DIR):
        subprocess.run(["git", "clone", "--depth", "1", REPO_URL, REPO_DIR], check=True)
    os.chdir(REPO_DIR)

import pandas as pd

df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

# Number 1: how many pages are currently declining
declining_rate = (df["trend_direction"].str.lower() == "down").mean()

# Number 2: how many pages are "stale + visible" -- reviewable candidates under a simple rule
stale_visible_mask = (df["days_since_last_update"] >= 180) & (df["impressions_90d"] >= 500)
stale_visible_count = stale_visible_mask.sum()
stale_visible_pct = stale_visible_mask.mean()

# Number 3: of those stale+visible candidates, how many are ALSO currently declining
declining_mask = df["trend_direction"].str.lower() == "down"
overlap_pct = (stale_visible_mask & declining_mask).sum() / stale_visible_count

print(f"Overall declining rate: {declining_rate:.1%}")
print(f"Stale + visible candidate pages: {stale_visible_count} ({stale_visible_pct:.1%} of {len(df)} rows)")
print(f"Of stale+visible candidates, {overlap_pct:.1%} are also currently declining")
print(f"\n-> A simple stale+visible rule alone doesn't fully separate declining from stable pages,")
print(f"   which is exactly the gap a learned model can close (as Weeks 1-2 already showed).")
```
