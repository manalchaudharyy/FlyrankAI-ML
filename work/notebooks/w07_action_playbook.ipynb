## 1. Ranked actions + reason codes

*The queue: what to do first, and why, in words a human trusts.*# Ranked Actions + Reason Codes

The model is intended to prioritize content pages that may require attention. The ranked queue supports content teams by identifying pages that should be reviewed first rather than automatically making decisions.

| Priority | Action | Reason Code |
|----------|--------|-------------|
| High | Refresh content immediately | HIGH_DECLINE_RISK |
| High | Review keyword targeting | LOW_CTR |
| Medium | Update metadata | LOW_IMPRESSIONS |
| Medium | Improve internal linking | TRAFFIC_DECAY |
| Low | Continue monitoring | STABLE_PAGE |

The queue should always be reviewed by a human before any content changes are made.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
import pandas as pd
import numpy as np

# Example queue (replace probability column with your model predictions if available)
queue = pd.DataFrame({
    "page_id": range(1,11),
    "decline_probability": np.round(np.random.uniform(0.45,0.95,10),2)
})

queue = queue.sort_values("decline_probability", ascending=False)

def assign_reason(p):
    if p >= 0.85:
        return "HIGH_DECLINE_RISK"
    elif p >= 0.70:
        return "LOW_CTR"
    elif p >= 0.55:
        return "TRAFFIC_DECAY"
    else:
        return "STABLE_PAGE"

queue["reason_code"] = queue["decline_probability"].apply(assign_reason)

queue
```
